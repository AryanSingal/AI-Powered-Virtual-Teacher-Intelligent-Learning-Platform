```python
import os

file_path = '/content/SadTalker/src/face3d/util/preprocess.py'
if os.path.exists(file_path):
    with open(file_path, 'r') as f:
        data = f.read()

    # Force 1D arrays to collapse into scalars using np.sum() before casting to float
    data = data.replace('float((t[0] - w0/2)*s)', 'float(np.sum((t[0] - w0/2)*s))')
    data = data.replace('float((h0/2 - t[1])*s)', 'float(np.sum((h0/2 - t[1])*s))')

    # Fix trans_params for NumPy 1.25+ compatibility
    old_line = 'trans_params = np.array([w0, h0, s, t[0], t[1]])'
    new_line = 'trans_params = np.array([float(np.sum(w0)), float(np.sum(h0)), float(np.sum(s)), float(np.sum(t[0])), float(np.sum(t[1]))])'

    if old_line in data:
        data = data.replace(old_line, new_line)

    with open(file_path, 'w') as f:
        f.write(data)
    print("✅ SadTalker array shape bug patched!")
```

```python
#STEP 6: Interface
#!/usr/bin/env python3
"""
GURU Virtual Tutor - Complete Platform with Fixed Layout & File Upload
All Issues Fixed: SadTalker working, Layout corrected, File upload for notes
"""

import os
import uuid
import json
import shutil
import subprocess
import time
import tempfile
import base64
import traceback
import numpy as np
import pandas as pd
from PIL import Image
import cv2
import gradio as gr
from datetime import datetime
import sys

# Optional AI imports with fallbacks
try:
    import whisper
except ImportError:
    whisper = None
    print("Warning: Whisper not available")

try:
    import docx
except ImportError:
    docx = None
    print("Warning: python-docx not available")

try:
    from pptx import Presentation
except ImportError:
    Presentation = None
    print("Warning: python-pptx not available")

try:
    from TTS.api import TTS
except ImportError:
    TTS = None
    print("Warning: TTS not available")

try:
    from pydub import AudioSegment
except ImportError:
    AudioSegment = None
    print("Warning: AudioSegment not available")

try:
    from ctransformers import AutoModelForCausalLM
    import torch  # Still import torch for the rest of your pipeline (SadTalker, TTS, etc.)
except ImportError:
    AutoModelForCausalLM = torch = None
    print("⚠️ Warning: cTransformers or Torch not available")


# =============================================================================
# CONFIGURATION & CONSTANTS
# =============================================================================
KB_PATH = "/content/knowledge_base.json"
TMP_DIR = "/content/tmp"
SADTALKER_ROOT = "/content/SadTalker"
LESSON_CONTENT_PATH = "/content/lesson_content.json"
DEFAULT_FACE_SIZE = 256
MAX_RETRIES = 3
GENERATION_TIMEOUT = 400

STATIC_DIR = "static"
os.makedirs(TMP_DIR, exist_ok=True)
os.makedirs(STATIC_DIR, exist_ok=True)
SHARED_FILES_DIR = "/content/shared_files"
FILE_REGISTRY_PATH = "/content/shared_files/file_registry.json"

os.makedirs(f"{SHARED_FILES_DIR}/notes", exist_ok=True)
os.makedirs(f"{SHARED_FILES_DIR}/assignments", exist_ok=True)
os.makedirs(f"{SHARED_FILES_DIR}/resources", exist_ok=True)

# File registry for tracking shared files
shared_files_registry = {"files": []}

def load_shared_files():
    """Load shared files registry"""
    global shared_files_registry
    if os.path.exists(FILE_REGISTRY_PATH):
        try:
            with open(FILE_REGISTRY_PATH, "r") as f:
                shared_files_registry = json.load(f)
        except:
            shared_files_registry = {"files": []}

def save_shared_files():
    """Save shared files registry"""
    try:
        with open(FILE_REGISTRY_PATH, "w") as f:
            json.dump(shared_files_registry, f, indent=2)
    except Exception as e:
        print(f"Error saving shared files: {e}")
# Load existing shared files
load_shared_files()

# User credentials
USERS = {
    "teacher": {"password": "teach123", "role": "teacher"},
    "student": {"password": "learn123", "role": "student"},
    "admin": {"password": "admin123", "role": "teacher"}
}

# =============================================================================
# GLOBAL STATE MANAGEMENT
# =============================================================================
current_user = {"username": None, "role": None, "logged_in": False}
teacher_profile = {"face": None, "audio": None}
video_library = []
student_videos = []
notes_data = []
quiz_data = []
uploaded_logo = None
last_generated_video = None

# Global variables for uploaded data
avatar_image_data = None
avatar_voice_data = None

# Knowledge base
knowledge_base = {}
if os.path.exists(KB_PATH):
    try:
        with open(KB_PATH, "r") as f:
            knowledge_base = json.load(f)
    except:
        knowledge_base = {}

def save_kb():
    try:
        with open(KB_PATH, "w") as f:
            json.dump(knowledge_base, f, indent=2)
    except Exception as e:
        print(f"Error saving knowledge base: {e}")

# Enhanced lesson content
lesson_content = {
    "notes": "👋 Hello students! In this platform you can:\n\n- View your lesson notes in the **Notes** tab.\n- Access shared files from your teacher in the **Downloads** tab.\n- Attempt your quiz questions in the **Quiz** tab.\n- Browse and play created videos in the **Center Video Player**.\n- Use the **Right Panel** to ask your doubts directly to the teacher.",

    "quiz": [
        {
            "question": "Which tab allows you to access the files shared by your teacher?",
            "options": ["A) Notes", "B) Download", "C) Quiz", "D) Extras"],
            "answer": "B) Download"
        }
    ],

    "assignment": "✍️ Assignment: Explore the Virtual Guru platform. Write down what features you found most useful (e.g., Notes, Downloads, Quiz, Video Player, Doubt Panel).",

    "resources": "",
    "extras": "",
    "flashcards": "",
    "tips": ""
}


def save_lesson_content():
    try:
        with open(LESSON_CONTENT_PATH, "w") as f:
            json.dump(lesson_content, f, indent=2)
    except Exception as e:
        print(f"Error saving lesson content: {e}")

def load_lesson_content():
    global lesson_content
    if os.path.exists(LESSON_CONTENT_PATH):
        try:
            with open(LESSON_CONTENT_PATH, "r") as f:
                lesson_content = json.load(f)
        except:
            pass

asked_questions_history = []
# Load existing lesson content
load_lesson_content()


def refresh_student_components():
    """Refresh student dashboard with latest lesson content"""
    load_lesson_content()  # Reload from JSON

    notes_update = gr.update(value=lesson_content["notes"])
    quiz_update = gr.update(value=f"Total Quiz Questions: {len(lesson_content['quiz'])}\n\n" +
                           "\n".join([f"Q{i+1}: {q['question']}\nAnswer: {q['answer']}\n"
                                    for i, q in enumerate(lesson_content['quiz'])]))
    assignment_update = gr.update(value=lesson_content["assignment"])
    resources_update = gr.update(value=lesson_content["resources"])
    extras_update = gr.update(value=lesson_content["extras"])
    flashcards_update = gr.update(value=lesson_content["flashcards"])
    tips_update = gr.update(value=lesson_content["tips"])

    return (notes_update, quiz_update, assignment_update,
            resources_update, extras_update, flashcards_update, tips_update)

# =============================================================================
# HELPER FUNCTIONS
# =============================================================================
def get_welcome_name(default="Student"):
    """Returns properly capitalized username for welcome display with fallback."""
    if not current_user or not isinstance(current_user, dict):
        return default
    username = current_user.get('username')
    if not username:
        username = default
    return str(username).title()

# Add this after LESSON_CONTENT_PATH
QUESTIONS_HISTORY_PATH = "/content/questions_history.json"

def save_questions_history():
    """Save question history to file"""
    try:
        with open(QUESTIONS_HISTORY_PATH, "w") as f:
            json.dump(asked_questions_history, f, indent=2)
        print(f"💾 Saved {len(asked_questions_history)} questions")
    except Exception as e:
        print(f"Error saving questions: {e}")

# STEP 3: Replace your existing file upload function with this
def share_file_with_students(uploaded_file, category="notes", description=""):
    """Share file with students - NO TEXT EXTRACTION"""
    try:
        if uploaded_file is None:
            return "❌ No file uploaded"

        print(f"📤 Sharing file: {uploaded_file.name}")

        # Generate unique filename
        timestamp = int(time.time())
        original_name = os.path.basename(uploaded_file.name)
        file_ext = os.path.splitext(original_name)[1]
        unique_filename = f"{timestamp}_{uuid.uuid4().hex[:8]}{file_ext}"

        # Save to shared directory
        category_path = os.path.join(SHARED_FILES_DIR, category)
        destination_path = os.path.join(category_path, unique_filename)

        # Copy file (PRESERVE ORIGINAL - NO TEXT EXTRACTION)
        shutil.copy2(uploaded_file.name, destination_path)

        if not os.path.exists(destination_path):
            return "❌ Failed to save file"

        # Get file size
        file_size = os.path.getsize(destination_path)
        if file_size < 1024:
            size_str = f"{file_size}B"
        elif file_size < 1024*1024:
            size_str = f"{file_size/1024:.1f}KB"
        else:
            size_str = f"{file_size/(1024*1024):.1f}MB"

        # Add to registry
        file_record = {
            "id": f"file_{timestamp}_{uuid.uuid4().hex[:8]}",
            "original_name": original_name,
            "stored_filename": unique_filename,
            "category": category,
            "upload_date": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
            "uploaded_by": current_user.get("username", "teacher"),
            "file_size": size_str,
            "description": description or f"{category.title()} file",
            "file_path": destination_path
        }

        shared_files_registry["files"].append(file_record)
        save_shared_files()

        return f"""
✅ **FILE SHARED SUCCESSFULLY**

📄 **File:** {original_name}
📁 **Category:** {category.title()}
📏 **Size:** {size_str}
👥 **Shared with:** All Students
🕒 **Shared:** {file_record['upload_date']}

Students can now download this file from their dashboard.
        """

    except Exception as e:
        print(f"❌ Error sharing file: {e}")
        return f"❌ Error sharing file: {str(e)}"

def get_shared_files(category=None):
    """Get shared files for display"""
    try:
        load_shared_files()  # Refresh from file
        files = shared_files_registry.get("files", [])

        if category:
            files = [f for f in files if f.get("category") == category]

        return files
    except Exception as e:
        print(f"Error getting shared files: {e}")
        return []

def delete_shared_file(file_id):
    """Delete a shared file"""
    try:
        files = shared_files_registry.get("files", [])

        for i, file_record in enumerate(files):
            if file_record.get("id") == file_id:
                # Remove from registry
                del shared_files_registry["files"][i]

                # Delete physical file
                file_path = file_record.get("file_path")
                if file_path and os.path.exists(file_path):
                    os.remove(file_path)

                save_shared_files()
                return f"✅ Deleted: {file_record.get('original_name')}"

        return "❌ File not found"

    except Exception as e:
        return f"❌ Error deleting file: {str(e)}"

def load_questions_history():
    """Load question history from file"""
    global asked_questions_history
    if os.path.exists(QUESTIONS_HISTORY_PATH):
        try:
            with open(QUESTIONS_HISTORY_PATH, "r") as f:
                asked_questions_history = json.load(f)
            print(f"📂 Loaded {len(asked_questions_history)} questions")
        except Exception as e:
            print(f"Error loading questions: {e}")
            asked_questions_history = []
    else:
        asked_questions_history = []

def safe_get_user_role():
    """Safely get user role with fallback"""
    if not current_user or not isinstance(current_user, dict):
        return "student"
    role = current_user.get('role')
    if not role:
        role = "student"
    return str(role).lower()
load_questions_history()
# =============================================================================
# MODEL LOADING
# =============================================================================
tokenizer = None
tiny_llama = None
coqui_tts = None
yourtts = None
whisper_model = None

print("🤖 Loading AI models...")

try:
    from ctransformers import AutoModelForCausalLM

    print("Loading TinyLLaMA (GGUF) model...")
    # Load the quantized GGUF model (download first if not present)
    tiny_llama = AutoModelForCausalLM.from_pretrained(
        ".",  # directory where your GGUF file is stored
        model_file="tinyllama-1.1b-chat-v1.0.Q4_K_M.gguf",
        model_type="llama",
        gpu_layers=20  # Adjust based on your Colab GPU VRAM
    )
    print("✅ TinyLLaMA (GGUF) loaded successfully")

except Exception as e:
    print(f"⚠️ Warning: TinyLLaMA (GGUF) not loaded - {e}")


try:
    from TTS.api import TTS
    print("Loading TTS models...")

    # Load default TTS model first
    coqui_tts = TTS("tts_models/en/ljspeech/tacotron2-DDC")
    print("✅ Default TTS model loaded")

    # Load YourTTS with proper model path and validation
    try:
        print("Loading YourTTS for voice cloning...")

        # Use the correct YourTTS model path
        yourtts = TTS("tts_models/multilingual/multi-dataset/your_tts", progress_bar=False)

        # Validate the model by checking required attributes
        if hasattr(yourtts, 'synthesizer') and yourtts.synthesizer:
            print("✅ YourTTS model loaded and validated")
        else:
            print("⚠️ YourTTS model loaded but may have compatibility issues")

    except Exception as e:
        print(f"❌ YourTTS loading failed: {e}")
        print("Trying alternative voice cloning models...")

        # Try alternative models for voice cloning
        alternative_models = [
            "tts_models/en/vctk/vits",
            "tts_models/en/ljspeech/vits--neon",
            "tts_models/multilingual/multi-dataset/bark"
        ]

        for alt_model in alternative_models:
            try:
                yourtts = TTS(alt_model, progress_bar=False)
                print(f"✅ Alternative voice cloning model loaded: {alt_model}")
                break
            except Exception as alt_e:
                print(f"❌ {alt_model} failed: {alt_e}")
                continue

        if not yourtts:
            print("⚠️ Voice cloning disabled - no compatible models found")

    print("✅ TTS initialization complete")

except ImportError:
    print("❌ TTS library not available")
except Exception as e:
    print(f"❌ TTS initialization failed: {e}")

try:
    if whisper:
        print("Loading Whisper model...")
        whisper_model = whisper.load_model("base")
        print("✅ Whisper loaded successfully")
except Exception as e:
    print(f"⚠️ Warning: Whisper not loaded - {e}")

print("🎨 Models loading complete!")

# =============================================================================
# ENHANCED CUSTOM CSS
# =============================================================================
custom_css = """
<style>
    /* Global Styles - Clean and Bright */
    .gradio-container {
        font-family: 'Inter', 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif !important;
        background: linear-gradient(135deg, #f0f4f8 0%, #e3f2fd 100%) !important;
        font-size: 16px !important;
        padding: 20px !important;
        color: #000000 !important;
    }

    /* Force Markdown and standard text to black */
    .prose, .prose * {
        color: #000000 !important;
    }

    /* Header Styling */
    .app-header {
        background: linear-gradient(135deg, #ffffff 0%, #f8fafc 50%, #e0f2fe 100%) !important;
        padding: 30px !important;
        border-radius: 20px !important;
        box-shadow: 0 4px 20px rgba(59, 130, 246, 0.1) !important;
        margin-bottom: 25px !important;
        border: 1px solid rgba(59, 130, 246, 0.1) !important;
        text-align: center !important;
    }

    .app-title {
        font-size: 3em !important;
        font-weight: 900 !important;
        color: #1e40af !important;
        margin: 0 !important;
        text-shadow: 0 2px 4px rgba(30, 64, 175, 0.1) !important;
    }

    .app-subtitle {
        background: linear-gradient(90deg, #dbeafe 0%, #bfdbfe 100%) !important;
        color: #1e40af !important;
        padding: 15px 25px !important;
        border-radius: 25px !important;
        font-weight: 600 !important;
        margin: 15px 0 !important;
        font-size: 1.2em !important;
        display: inline-block !important;
    }

    /* Three Column Panel Styles */
    .left-panel, .center-panel, .right-panel {
        flex: 1 !important;
        display: flex !important;
        flex-direction: column !important;
        justify-content: flex-start !important;
        align-items: stretch !important;
        height: auto !important;
        box-sizing: border-box !important;
        border-radius: 20px !important;
        padding: 25px !important;
        margin: 10px !important;
        box-shadow: 0 6px 20px rgba(0, 0, 0, 0.1) !important;
        border: 2px solid rgba(0, 0, 0, 0.1) !important;
        position: relative !important;
        overflow-y: visible !important;
        background-clip: padding-box !important;
    }

    /* Equal height rows */
    .gr-block.gr-row {
        display: flex !important;
        align-items: stretch !important;
    }

    .left-panel {
        background: linear-gradient(135deg, #fff4e6 0%, #ffeaa7 50%, #fdcb6e 100%) !important;
        border-color: rgba(253, 203, 110, 0.3) !important;
        box-shadow: 0 6px 20px rgba(253, 203, 110, 0.2) !important;
    }

    .left-panel::before {
        content: '' !important;
        position: absolute !important;
        top: 0 !important;
        left: 0 !important;
        right: 0 !important;
        height: 6px !important;
        background: linear-gradient(90deg, #e17055, #fdcb6e, #f39c12) !important;
    }

    #left-panel-container {
         display: flex;
         flex-direction: column;
         justify-content: flex-start !important;
         height: 100%;
    }

    .center-panel {
        background: linear-gradient(135deg, #e8f5e8 0%, #a8e6a3 50%, #81c784 100%) !important;
        border-color: rgba(129, 199, 132, 0.3) !important;
        box-shadow: 0 6px 20px rgba(129, 199, 132, 0.2) !important;
    }

    .center-panel::before {
        content: '' !important;
        position: absolute !important;
        top: 0 !important;
        left: 0 !important;
        right: 0 !important;
        height: 6px !important;
        background: linear-gradient(90deg, #4caf50, #81c784, #66bb6a) !important;
    }

    .right-panel {
        background: linear-gradient(135deg, #e3f2fd 0%, #90caf9 50%, #64b5f6 100%) !important;
        border-color: rgba(100, 181, 246, 0.3) !important;
        box-shadow: 0 6px 20px rgba(100, 181, 246, 0.2) !important;
    }

    .right-panel::before {
        content: '' !important;
        position: absolute !important;
        top: 0 !important;
        left: 0 !important;
        right: 0 !important;
        height: 6px !important;
        background: linear-gradient(90deg, #2196f3, #64b5f6, #42a5f5) !important;
    }

    /* Section Headers */
    .section-header {
        font-size: 1.8em !important;
        font-weight: 800 !important;
        color: #1e293b !important;
        margin: 20px 0 15px 0 !important;
        text-align: center !important;
        text-transform: none !important;
        letter-spacing: 0.5px !important;
    }

    /* All subsection headers */
    .left-panel h3, .center-panel h3, .right-panel h3 {
        font-weight: 800 !important;
    }

    /* Enhanced Button Styles */
    .primary-button {
        background: linear-gradient(135deg, #3b82f6 0%, #2563eb 50%, #1d4ed8 100%) !important;
        color: white !important;
        border: none !important;
        border-radius: 12px !important;
        padding: 14px 28px !important;
        font-weight: 700 !important;
        font-size: 1.1em !important;
        cursor: pointer !important;
        transition: all 0.3s ease !important;
        box-shadow: 0 4px 15px rgba(59, 130, 246, 0.3) !important;
        margin: 10px auto !important;
        display: block !important;
        min-width: 150px !important;
    }

    .primary-button:hover {
        transform: translateY(-2px) scale(1.05) !important;
        box-shadow: 0 6px 20px rgba(59, 130, 246, 0.4) !important;
        background: linear-gradient(135deg, #2563eb 0%, #1d4ed8 50%, #1e40af 100%) !important;
    }

    .upload-button {
        background: linear-gradient(135deg, #10b981 0%, #059669 50%, #047857 100%) !important;
        color: white !important;
        border: none !important;
        border-radius: 12px !important;
        padding: 12px 24px !important;
        font-weight: 700 !important;
        font-size: 1em !important;
        cursor: pointer !important;
        transition: all 0.3s ease !important;
        box-shadow: 0 4px 15px rgba(16, 185, 129, 0.3) !important;
        margin: 10px auto !important;
        display: block !important;
        min-width: 140px !important;
    }

    .upload-button:hover {
        transform: translateY(-2px) scale(1.05) !important;
        box-shadow: 0 6px 20px rgba(16, 185, 129, 0.4) !important;
    }

    .logout-button {
        background: linear-gradient(135deg, #f59e0b 0%, #d97706 50%, #b45309 100%) !important;
        color: white !important;
        border: none !important;
        border-radius: 12px !important;
        padding: 10px 20px !important;
        font-weight: 700 !important;
        font-size: 1em !important;
        cursor: pointer !important;
        transition: all 0.3s ease !important;
        box-shadow: 0 4px 15px rgba(245, 158, 11, 0.3) !important;
    }

    .logout-button:hover {
        transform: translateY(-2px) scale(1.05) !important;
        box-shadow: 0 6px 20px rgba(245, 158, 11, 0.4) !important;
    }

    /* Enhanced Input Styles */
    /* Enhanced Input Styles */
    .gradio-textbox input, .gradio-textbox textarea {
        background: #e0f2fe !important; /* Light blue background for readability */
        border: 2px solid rgba(59, 130, 246, 0.4) !important;
        border-radius: 12px !important;
        padding: 16px !important;
        font-size: 1.1em !important;
        font-weight: 500 !important;
        color: #000000 !important; /* Pure black text */
        transition: all 0.3s ease !important;
        resize: vertical !important;
    }

    .gradio-textbox input::placeholder, .gradio-textbox textarea::placeholder {
        font-size: 1.1em !important;
        font-weight: 600 !important;
        color: #6b7280 !important;
        opacity: 0.8 !important;
    }

    .gradio-textbox input:focus, .gradio-textbox textarea:focus {
        border-color: #3b82f6 !important;
        box-shadow: 0 0 0 4px rgba(59, 130, 246, 0.1) !important;
        outline: none !important;
    }

    /* Enhanced Labels */
    .gradio-textbox label, .gradio-file label, .gradio-image label, .gradio-audio label, .gradio-dropdown label {
        font-size: 1.2em !important;
        font-weight: 700 !important;
        color: #1e293b !important;
        margin-bottom: 8px !important;
    }

    /* Dropdown Styles */
    .gradio-dropdown select {
        background: linear-gradient(135deg, rgba(255, 255, 255, 0.95) 0%, rgba(248, 250, 252, 0.9) 100%) !important;
        border: 2px solid rgba(59, 130, 246, 0.2) !important;
        border-radius: 12px !important;
        padding: 14px !important;
        font-size: 1.1em !important;
        font-weight: 600 !important;
        color: #000000 !important;
    }

    /* File Upload Styles */
    .gradio-file {
        background: linear-gradient(135deg, rgba(255, 255, 255, 0.9) 0%, rgba(248, 250, 252, 0.8) 100%) !important;
        border: 3px dashed #3b82f6 !important;
        border-radius: 15px !important;
        padding: 25px !important;
        transition: all 0.3s ease !important;
        min-height: 100px !important;
    }

    .gradio-file:hover {
        background: linear-gradient(135deg, rgba(239, 246, 255, 0.9) 0%, rgba(219, 234, 254, 0.8) 100%) !important;
        border-color: #2563eb !important;
        transform: translateY(-2px) !important;
        box-shadow: 0 6px 20px rgba(59, 130, 246, 0.15) !important;
    }

    /* Image Upload Areas */
    .gradio-image {
        border-radius: 15px !important;
        overflow: hidden !important;
        border: 2px solid rgba(59, 130, 246, 0.2) !important;
    }

    .gradio-audio {
        border-radius: 15px !important;
        overflow: hidden !important;
        border: 2px solid rgba(59, 130, 246, 0.2) !important;
    }

    /* Tab Styles */
    .gradio-tabs .tab-nav button {
        background: linear-gradient(135deg, #ffffff 0%, #f8fafc 100%) !important;
        color: #000000 !important; /* Forces unselected tab text to pure black */
        border: 2px solid rgba(59, 130, 246, 0.2) !important;
        border-radius: 10px 10px 0 0 !important;
        padding: 14px 22px !important;
        font-weight: 900 !important;
        font-size: 1.1em !important;
        margin-right: 4px !important;
        transition: all 0.3s ease !important;
    }

    .gradio-tabs .tab-nav button.selected {
        background: linear-gradient(135deg, #3b82f6 0%, #2563eb 100%) !important;
        color: #000000 !important; /* Forces selected tab text to pure black */
        transform: translateY(-2px) !important;
        box-shadow: 0 4px 15px rgba(59, 130, 246, 0.3) !important;
    }

    /* Video and HTML Container Styles */
    .video-container {
        border: 3px solid #3b82f6 !important;
        border-radius: 15px !important;
        padding: 20px !important;
        background: linear-gradient(135deg, #ffffff 0%, #f8fafc 100%) !important;
        box-shadow: 0 6px 20px rgba(59, 130, 246, 0.1) !important;
        margin: 15px 0 !important;
    }

    /* Login Container */
    .login-container {
        background: linear-gradient(135deg, rgba(255, 255, 255, 0.95) 0%, rgba(248, 250, 252, 0.9) 100%) !important;
        border-radius: 25px !important;
        padding: 50px !important;
        box-shadow: 0 15px 35px rgba(59, 130, 246, 0.15) !important;
        border: 2px solid rgba(59, 130, 246, 0.1) !important;
        max-width: 450px !important;
        margin: 0 auto !important;
    }

    /* Splash Screen */
    .splash-screen {
        background: linear-gradient(135deg, #f0f4f8 0%, #e3f2fd 50%, #e1f5fe 100%) !important;
        height: 100vh !important;
        display: flex !important;
        align-items: center !important;
        justify-content: center !important;
        flex-direction: column !important;
        position: fixed !important;
        top: 0 !important;
        left: 0 !important;
        width: 100% !important;
        z-index: 9999 !important;
    }

    .logo-container {
        text-align: center !important;
        padding: 40px !important;
        background: linear-gradient(135deg, rgba(255, 255, 255, 0.9) 0%, rgba(248, 250, 252, 0.8) 100%) !important;
        border-radius: 25px !important;
        box-shadow: 0 15px 35px rgba(59, 130, 246, 0.15) !important;
        border: 2px solid rgba(59, 130, 246, 0.1) !important;
    }

    .spinner {
        border: 8px solid #e3f2fd;
        border-top: 8px solid #3b82f6;
        border-radius: 50%;
        width: 60px;
        height: 60px;
        animation: spin 1s linear infinite;
        margin: 20px auto;
    }

    @keyframes spin {
        100% { transform: rotate(360deg); }
    }

    /* Responsive */
    @media (max-width: 768px) {
        .gradio-container {
            padding: 10px !important;
        }

        .left-panel, .center-panel, .right-panel {
            margin: 5px !important;
            padding: 15px !important;
        }

        .app-title {
            font-size: 2em !important;
        }
    }
</style>
"""

# =============================================================================
# AUTHENTICATION FUNCTIONS
# =============================================================================
def authenticate_user(username, password):
    """Authenticate user credentials"""
    if username in USERS and USERS[username]["password"] == password:
        current_user["username"] = username
        current_user["role"] = USERS[username]["role"]
        current_user["logged_in"] = True
        load_lesson_content()
        print(f"🔄 Content refreshed for {username}")
        return True, f"✅ Welcome, {username.title()}! Logged in as {USERS[username]['role'].title()}"
    return False, "❌ Invalid username or password"

def logout_user():
    """Logout current user"""
    current_user["username"] = None
    current_user["role"] = None
    current_user["logged_in"] = False
    return "👋 Logged out successfully"

def check_access(required_role="student"):
    """Check if current user has required access"""
    if not current_user["logged_in"]:
        return False
    if required_role == "teacher":
        return current_user["role"] == "teacher"
    return True

# =============================================================================
# UTILITY FUNCTIONS - FIXED IMAGE PROCESSING
# =============================================================================

def auto_resize_face_image(img, target_size=DEFAULT_FACE_SIZE):
    """Face image processing with full error handling"""
    try:
        print(f"📸 Processing face image of type: {type(img)}")

        if img is None:
            print("❌ Input image is None")
            return None

        # Handle PIL Image directly (most common case from Gradio)
        if hasattr(img, 'size') and hasattr(img, 'mode'):
            print(f"✅ PIL Image detected - Size: {img.size}, Mode: {img.mode}")
            im = img.copy()
        # Handle numpy array
        elif isinstance(img, np.ndarray):
            print(f"🔄 Converting numpy array to PIL - Shape: {img.shape}")
            if len(img.shape) == 3:
                im = Image.fromarray(img.astype(np.uint8))
            else:
                print("❌ Invalid numpy array shape")
                return None
        # Handle file path
        elif isinstance(img, str) and os.path.exists(img):
            print(f"📂 Loading image from path: {img}")
            im = Image.open(img)
        else:
            print(f"❌ Unsupported image type: {type(img)}")
            return None

        # Ensure RGB/RGBA mode
        if im.mode not in ['RGB', 'RGBA']:
            print(f"🎨 Converting from {im.mode} to RGB")
            im = im.convert('RGB')

        print(f"📏 Image size: {im.size}")

        # Make square by center cropping
        w, h = im.size
        if w != h:
            print("✂️ Cropping to square")
            size = min(w, h)
            left = (w - size) // 2
            top = (h - size) // 2
            im = im.crop((left, top, left + size, top + size))
            print(f"📐 Cropped size: {im.size}")

        # Resize to target
        if im.size[0] != target_size:
            print(f"🔄 Resizing to {target_size}x{target_size}")
            im = im.resize((target_size, target_size), Image.Resampling.LANCZOS)

        print(f"✅ Face processing complete - Final size: {im.size}")
        return im

    except Exception as e:
        print(f"❌ Face processing error: {str(e)}")
        traceback.print_exc()
        return None

def auto_resize_side_image(img, target_height):
    """Side image processing with full error handling"""
    try:
        print(f"🖼️ Processing side image - Target height: {target_height}")

        if img is None:
            print("❌ Side image is None")
            return None

        # Handle PIL Image directly
        if hasattr(img, 'size') and hasattr(img, 'mode'):
            print(f"✅ PIL Image detected - Size: {img.size}, Mode: {img.mode}")
            im = img.copy()
        # Handle numpy array
        elif isinstance(img, np.ndarray):
            print(f"🔄 Converting numpy array to PIL - Shape: {img.shape}")
            if len(img.shape) == 3:
                im = Image.fromarray(img.astype(np.uint8))
            else:
                print("❌ Invalid numpy array shape")
                return None
        # Handle file path
        elif isinstance(img, str) and os.path.exists(img):
            print(f"📂 Loading side image from path: {img}")
            im = Image.open(img)
        else:
            print(f"❌ Unsupported side image type: {type(img)}")
            return None

        # Ensure RGB mode
        if im.mode != 'RGB':
            print(f"🎨 Converting from {im.mode} to RGB")
            im = im.convert('RGB')

        print(f"📏 Original side image size: {im.size}")

        # Resize maintaining aspect ratio
        if im.height != target_height:
            ratio = target_height / im.height
            new_width = int(im.width * ratio)
            print(f"🔄 Resizing to {new_width}x{target_height}")
            im = im.resize((new_width, target_height), Image.Resampling.LANCZOS)

        print(f"✅ Side image processing complete - Final size: {im.size}")
        return im

    except Exception as e:
        print(f"❌ Side image processing error: {str(e)}")
        traceback.print_exc()
        return None

def save_image(img, prefix="img"):
    """Image saving with proper error handling"""
    try:
        if img is None:
            print("❌ Cannot save None image")
            return None

        # Generate unique filename with timestamp
        timestamp = int(time.time())
        filename = f"{prefix}_{timestamp}_{uuid.uuid4().hex[:8]}.png"
        path = os.path.join(TMP_DIR, filename)

        print(f"💾 Saving image to: {path}")

        # Save as PNG for best compatibility
        img.save(path, "PNG", optimize=True)

        # Verify file was created
        if os.path.exists(path) and os.path.getsize(path) > 0:
            file_size = os.path.getsize(path)
            print(f"✅ Image saved successfully: {file_size} bytes")
            return path
        else:
            print("❌ Image file not created or empty")
            return None

    except Exception as e:
        print(f"❌ Image save error: {str(e)}")
        traceback.print_exc()
        return None

def ensure_file_from_audio(audio):
    """Audio file validation"""
    try:
        if audio is None:
            print("❌ Audio is None")
            return None

        if isinstance(audio, str) and os.path.exists(audio):
            file_size = os.path.getsize(audio)
            print(f"✅ Audio file found: {audio} ({file_size} bytes)")
            return audio
        else:
            print(f"❌ Audio file not found: {audio}")
            return None

    except Exception as e:
        print(f"❌ Audio validation error: {str(e)}")
        return None

def get_video_dims(video_path):
    """Get video dimensions using OpenCV"""
    try:
        cap = cv2.VideoCapture(video_path)
        if not cap.isOpened():
            print(f"❌ Cannot open video: {video_path}")
            return DEFAULT_FACE_SIZE, DEFAULT_FACE_SIZE

        width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
        height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
        cap.release()

        print(f"📹 Video dimensions: {width}x{height}")
        return width, height

    except Exception as e:
        print(f"❌ Video dimension error: {str(e)}")
        return DEFAULT_FACE_SIZE, DEFAULT_FACE_SIZE

def has_video_stream(video_path):
    """Check if video file has valid video stream"""
    try:
        if not os.path.exists(video_path):
            return False

        if os.path.getsize(video_path) == 0:
            return False

        # Quick check with OpenCV
        cap = cv2.VideoCapture(video_path)
        if not cap.isOpened():
            return False

        ret, frame = cap.read()
        cap.release()

        return ret and frame is not None

    except Exception as e:
        print(f"❌ Video stream check error: {str(e)}")
        return False

# =============================================================================
# AI SERVICES - WORKING WITH SADTALKER
# =============================================================================
def safe_llm_answer(question):
    """Generate AI answer using GGUF TinyLLaMA or fallback - ends on complete sentences"""
    try:
        if tiny_llama:  # ctransformers model
            prompt = f"<|user|>\n{question}\n<|assistant|>\n"
            raw_answer = tiny_llama(prompt, max_new_tokens=25, temperature=0.7).strip()

            def truncate_to_last_sentence(text):
                for ending in ['. ', '! ', '? ']:
                    last_pos = text.rfind(ending)
                    if last_pos != -1:
                        return text[:last_pos + 1]
                if text.endswith(('.', '!', '?')):
                    return text
                return text

            return truncate_to_last_sentence(raw_answer)

        # Fallback if model is missing
        return f"Let me explain about {question}: This is an important educational topic."

    except Exception as e:
        print(f"❌ LLM error: {str(e)}")
        return f"Let me explain about {question}: This is an important subject that requires careful study."




def safe_tts_synthesis(text, voice_audio=None):
    """Enhanced TTS with better voice cloning completion handling"""
    output_path = os.path.join(TMP_DIR, f"tts_{uuid.uuid4().hex[:8]}.wav")

    text = text.strip() or "Processing request."
    print(f"🔊 Generating TTS: {text[:50]}...")

    # Enhanced voice cloning with better completion handling
    if voice_audio and yourtts and os.path.exists(voice_audio):
        try:
            print("🎤 Attempting voice cloning...")

            # Method 1: Standard approach with validation
            yourtts.tts_to_file(
                text=text,
                speaker_wav=voice_audio,
                file_path=output_path,
                language="en"
            )

            # Enhanced validation - wait for file and check size
            import time
            max_wait = 10  # seconds
            wait_time = 0

            while wait_time < max_wait:
                if os.path.exists(output_path):
                    file_size = os.path.getsize(output_path)
                    if file_size > 1000:  # At least 1KB
                        print(f"✅ Voice cloning successful (size: {file_size} bytes)")
                        return output_path
                    elif file_size > 0:
                        # File exists but might still be writing
                        time.sleep(0.5)
                        wait_time += 0.5
                        continue
                    else:
                        # Empty file, break and try alternatives
                        break
                else:
                    # File doesn't exist yet, wait a bit
                    time.sleep(0.5)
                    wait_time += 0.5

            print("Voice cloning processed but output validation failed")

        except Exception as e:
            print(f"Voice cloning error: {e}")

    # Try default TTS (your existing working code)
    try:
        if coqui_tts:
            print("🔊 Using default TTS...")
            coqui_tts.tts_to_file(text=text, file_path=output_path)
            if os.path.exists(output_path) and os.path.getsize(output_path) > 0:
                print("✅ Default TTS successful")
                return output_path
    except Exception as e:
        print(f"Default TTS error: {e}")

    # Fallback audio (your existing working code)
    print("🎵 Generating fallback audio...")
    sample_rate = 22050
    duration = max(2, len(text) * 0.08)
    t = np.linspace(0, duration, int(sample_rate * duration))
    audio_data = 0.1 * np.sin(440 * 2 * np.pi * t).astype(np.float32)

    try:
        from scipy.io.wavfile import write
        write(output_path, sample_rate, (audio_data * 32767).astype(np.int16))
        print("✅ Fallback audio generated")
    except Exception as e:
        print(f"Fallback audio error: {e}")

    return output_path



def generate_talking_video_safe(face_path, wav_path):
    """SadTalker video generation with proper error handling"""
    try:
        print("🎬 Starting SadTalker video generation...")
        print(f"   Face image: {face_path}")
        print(f"   Audio file: {wav_path}")

        # Validate inputs
        if not face_path or not os.path.exists(face_path):
            return "❌ Please set up your teacher profile first...", None, None


        if not wav_path or not os.path.exists(wav_path):
            print("❌ Audio file not found")
            return None

        # Check SadTalker installation
        if not os.path.exists(SADTALKER_ROOT):
            print(f"❌ SadTalker not found at: {SADTALKER_ROOT}")
            print("Please install SadTalker first:")
            print("git clone https://github.com/Winfredy/SadTalker.git /content/SadTalker")
            return None

        # Create output directory
        output_dir = os.path.join(TMP_DIR, f"sadtalker_{int(time.time())}_{uuid.uuid4().hex[:8]}")
        os.makedirs(output_dir, exist_ok=True)

        # Copy input files to SadTalker expected locations
        face_input = os.path.join(output_dir, "input_face.png")
        audio_input = os.path.join(output_dir, "input_audio.wav")

        try:
            shutil.copy2(face_path, face_input)
            shutil.copy2(wav_path, audio_input)
            print(f"✅ Input files copied to SadTalker directory")
        except Exception as e:
            print(f"❌ Failed to copy input files: {str(e)}")
            return None

        # Construct SadTalker command
        cmd = [
            'python', 'inference.py',
            '--driven_audio', audio_input,
            '--source_image', face_input,
            '--result_dir', output_dir,
            '--preprocess', 'full',
            '--size', '256',
            '--expression_scale', '0.7',
            '--still'
        ]

        print(f"🚀 Running SadTalker command...")
        print(f"   Command: {' '.join(cmd)}")

        # Set up environment
        env = os.environ.copy()
        env['CUDA_VISIBLE_DEVICES'] = '0'

        # Run SadTalker
        try:
            result = subprocess.run(
                cmd,
                cwd=SADTALKER_ROOT,
                capture_output=True,
                text=True,
                timeout=GENERATION_TIMEOUT,
                env=env
            )

            print(f"📊 SadTalker return code: {result.returncode}")

            if result.stdout:
                print(f"📝 SadTalker stdout: {result.stdout[-500:]}")
            if result.stderr:
                print(f"📝 SadTalker stderr: {result.stderr[-500:]}")

        except subprocess.TimeoutExpired:
            print(f"❌ SadTalker timed out after {GENERATION_TIMEOUT} seconds")
            return None
        except Exception as e:
            print(f"❌ SadTalker execution error: {str(e)}")
            return None

        # Look for generated video
        if result.returncode == 0:
            print("✅ SadTalker completed, searching for output video...")

            # Search recursively for MP4 files
            for root, dirs, files in os.walk(output_dir):
                for file in files:
                    if file.endswith('.mp4'):
                        video_path = os.path.join(root, file)
                        print(f"🔍 Found video: {video_path}")

                        if has_video_stream(video_path):
                            file_size = os.path.getsize(video_path)
                            print(f"✅ Valid video generated: {file_size} bytes")
                            return video_path
                        else:
                            print(f"❌ Invalid video stream: {video_path}")

            print("❌ No valid video found in output")
            print(f"📁 Output directory contents: {os.listdir(output_dir)}")
        else:
            print(f"❌ SadTalker failed with return code: {result.returncode}")

        return None

    except Exception as e:
        print(f"❌ SadTalker generation error: {str(e)}")
        traceback.print_exc()
        return None

def fast_side_image_stack(video_path, side_img, out_path):
    """Side image stacking with proper FFmpeg handling"""
    try:
        print("🖼️ Starting side image stacking...")
        print(f"   Video: {video_path}")
        print(f"   Output: {out_path}")

        if not video_path or not os.path.exists(video_path):
            print("❌ Video file not found")
            return

        if side_img is None:
            print("❌ No side image provided, copying original")
            shutil.copy2(video_path, out_path)
            return

        # Get video dimensions
        vid_width, vid_height = get_video_dims(video_path)
        print(f"📹 Video dimensions: {vid_width}x{vid_height}")

        # Process side image
        side_resized = auto_resize_side_image(side_img, vid_height)
        if not side_resized:
            print("❌ Failed to process side image")
            shutil.copy2(video_path, out_path)
            return

        side_width = side_resized.width
        print(f"🖼️ Side image dimensions: {side_width}x{vid_height}")

        # Ensure even dimensions for FFmpeg
        if (vid_width + side_width) % 2 != 0:
            print("🔧 Adjusting width for FFmpeg compatibility")
            padded = Image.new('RGB', (side_width + 1, vid_height), (255, 255, 255))
            padded.paste(side_resized, (0, 0))
            side_resized = padded
            side_width += 1

        # Save side image temporarily
        side_temp_path = save_image(side_resized, "side_temp")
        if not side_temp_path:
            print("❌ Failed to save side image")
            shutil.copy2(video_path, out_path)
            return

        # Get video duration for side image loop
        try:
            cap = cv2.VideoCapture(video_path)
            fps = cap.get(cv2.CAP_PROP_FPS)
            frame_count = cap.get(cv2.CAP_PROP_FRAME_COUNT)
            duration = frame_count / fps if fps > 0 else 5.0
            cap.release()
        except:
            duration = 5.0

        print(f"⏱️ Video duration: {duration:.2f} seconds")

        # FFmpeg command for side stacking
        cmd = [
            'ffmpeg', '-y',
            '-i', video_path,
            '-loop', '1', '-i', side_temp_path,
            '-filter_complex',
            f'[1:v]scale={side_width}:{vid_height}[side];[0:v][side]hstack=inputs=2',
            '-t', str(duration),
            '-c:v', 'libx264',
            '-preset', 'ultrafast',
            '-crf', '23',
            '-pix_fmt', 'yuv420p',
            '-c:a', 'aac',
            '-b:a', '128k',
            '-shortest',
            out_path
        ]

        print("🎬 Running FFmpeg for side stacking...")

        try:
            result = subprocess.run(
                cmd,
                capture_output=True,
                text=True,
                timeout=60
            )

            if result.returncode == 0:
                if os.path.exists(out_path) and os.path.getsize(out_path) > 0:
                    print(f"✅ Side stacking successful: {os.path.getsize(out_path)} bytes")
                else:
                    print("❌ Output file empty or not created")
                    shutil.copy2(video_path, out_path)
            else:
                print(f"❌ FFmpeg failed: {result.stderr}")
                shutil.copy2(video_path, out_path)

        except subprocess.TimeoutExpired:
            print("❌ FFmpeg timeout")
            shutil.copy2(video_path, out_path)
        except Exception as e:
            print(f"❌ FFmpeg error: {str(e)}")
            shutil.copy2(video_path, out_path)

        # Clean up temp file
        try:
            if os.path.exists(side_temp_path):
                os.remove(side_temp_path)
        except:
            pass

    except Exception as e:
        print(f"❌ Side stacking error: {str(e)}")
        if video_path and out_path:
            shutil.copy2(video_path, out_path)

def add_subtitles(video_path, text, out_path):
    """Add subtitles to video using FFmpeg"""
    try:
        print("📝 Adding subtitles...")

        if not video_path or not os.path.exists(video_path):
            return

        if not text or not text.strip():
            shutil.copy2(video_path, out_path)
            return

        # Create subtitle file
        srt_path = os.path.join(TMP_DIR, f"subtitle_{int(time.time())}.srt")

        # Get video duration
        try:
            cap = cv2.VideoCapture(video_path)
            fps = cap.get(cv2.CAP_PROP_FPS)
            frame_count = cap.get(cv2.CAP_PROP_FRAME_COUNT)
            duration = int(frame_count / fps) if fps > 0 else 5
            cap.release()
        except:
            duration = 5

        # Create SRT content
        srt_content = f"1\n00:00:00,000 --> 00:00:{duration:02d},000\n{text.strip()}\n"

        with open(srt_path, 'w', encoding='utf-8') as f:
            f.write(srt_content)

        # FFmpeg command for subtitles
        cmd = [
            'ffmpeg', '-y',
            '-i', video_path,
            '-vf', f'subtitles={srt_path}',
            '-c:v', 'libx264',
            '-preset', 'ultrafast',
            '-crf', '23',
            '-c:a', 'copy',
            out_path
        ]

        try:
            result = subprocess.run(cmd, capture_output=True, text=True, timeout=60)

            if result.returncode == 0:
                print("✅ Subtitles added successfully")
            else:
                print(f"❌ Subtitle FFmpeg failed: {result.stderr}")
                shutil.copy2(video_path, out_path)

        except Exception as e:
            print(f"❌ Subtitle error: {str(e)}")
            shutil.copy2(video_path, out_path)

        # Clean up
        try:
            if os.path.exists(srt_path):
                os.remove(srt_path)
        except:
            pass

    except Exception as e:
        print(f"❌ Subtitle processing error: {str(e)}")
        if video_path and out_path:
            shutil.copy2(video_path, out_path)



def add_question_to_history(question):
    global asked_questions_history
    if question and question.strip():
        clean_q = question.strip()
        if clean_q not in asked_questions_history:
            asked_questions_history.append(clean_q)
            if len(asked_questions_history) > 25:  # Keep last 25
                asked_questions_history = asked_questions_history[-25:]
            save_questions_history()
def get_question_choices():
    if not asked_questions_history:
        return ["No questions asked yet"]
    choices = []
    for i, q in enumerate(reversed(asked_questions_history[-10:]), 1):
        text = f"Q{i}: {q[:45]}..." if len(q) > 45 else f"Q{i}: {q}"
        choices.append(text)
    return choices

def select_from_dropdown(selected):
    if not selected or "No questions" in selected:
        return ""
    if ": " in selected:
        part = selected.split(": ", 1)[1]
        if part.endswith("..."):
            # Find full question
            search = part[:-3]
            for q in asked_questions_history:
                if q.startswith(search):
                    return q
        return part
    return selected
# =============================================================================
# LOGO FUNCTIONS
# =============================================================================
def create_logo():
    """Create logo with uploaded image or fallback to owl"""
    if uploaded_logo is not None:
        try:
            logo_save_path = os.path.join(STATIC_DIR, f"logo_{uuid.uuid4().hex[:8]}.png")
            uploaded_logo.save(logo_save_path, "PNG")

            return f"""
            <div style="text-align: center; margin: 20px 0;">
                <img src="/file={logo_save_path}" style="width: 80px; height: 80px; border-radius: 50%; margin-bottom: 15px; border: 3px solid #1e40af;">
                <div style="font-size: 2.2em; font-weight: 900; color: #1e40af; text-shadow: 0 2px 4px rgba(30, 64, 175, 0.2);">GURU</div>
                <div style="font-size: 1.4em; color: #64748b; font-weight: 600; margin-top: 5px;">Virtual Tutor</div>
            </div>
            """
        except Exception as e:
            print(f"Logo creation error: {e}")

    # Fallback to owl emoji
    return """
    <div style="text-align: center; margin: 20px 0;">
        <div style="font-size: 4.5em; margin-bottom: 15px;">🦉</div>
        <div style="font-size: 2.2em; font-weight: 900; color: #1e40af; text-shadow: 0 2px 4px rgba(30, 64, 175, 0.2);">GURU</div>
        <div style="font-size: 1.4em; color: #64748b; font-weight: 600; margin-top: 5px;">Virtual Tutor</div>
    </div>
    """

# =============================================================================
# TEACHER FUNCTIONS
# =============================================================================

def upload_avatar_image(image):
    """Handle avatar image upload"""
    global avatar_image_data
    try:
        if image is not None:
            avatar_image_data = image
            print(f"✅ Avatar image uploaded: {type(image)}, Size: {image.size if hasattr(image, 'size') else 'Unknown'}")
            return "✅ Avatar image uploaded successfully!"
        return "⚠️ Please select an image"
    except Exception as e:
        print(f"❌ Avatar upload error: {str(e)}")
        return f"❌ Error uploading image: {str(e)}"

def upload_avatar_voice(audio):
    """Handle avatar voice upload"""
    global avatar_voice_data
    try:
        if audio is not None:
            avatar_voice_data = audio
            print(f"✅ Avatar voice uploaded: {audio}")
            return "✅ Avatar voice uploaded successfully!"
        return "⚠️ Please select an audio file"
    except Exception as e:
        print(f"❌ Voice upload error: {str(e)}")
        return f"❌ Error uploading voice: {str(e)}"

def set_profile():
    """Set teacher profile with uploaded avatar and voice"""
    global avatar_image_data, avatar_voice_data, teacher_profile

    try:
        if not check_access("teacher"):
            return "❌ Access denied. Teacher login required."

        print("🔧 Setting teacher profile...")
        print(f"   Image data: {type(avatar_image_data)}")
        print(f"   Voice data: {avatar_voice_data}")

        if avatar_image_data is None:
            return "❌ Please upload avatar image first"

        if avatar_voice_data is None:
            return "❌ Please upload avatar voice first"

        # Process face image
        print("📸 Processing face image...")
        processed_face = auto_resize_face_image(avatar_image_data)
        if not processed_face:
            return "❌ Failed to process face image"

        # Save face image
        face_path = save_image(processed_face, "teacher_face")
        if not face_path:
            return "❌ Failed to save face image"

        # Validate voice audio
        voice_path = ensure_file_from_audio(avatar_voice_data)
        if not voice_path:
            return "❌ Invalid voice audio file"

        # Update teacher profile
        teacher_profile["face"] = face_path
        teacher_profile["audio"] = voice_path

        print(f"✅ Profile set successfully!")
        print(f"   Face: {face_path}")
        print(f"   Voice: {voice_path}")

        return "🎉 Teacher profile set successfully! You can now generate videos."

    except Exception as e:
        print(f"❌ Profile setup error: {str(e)}")
        traceback.print_exc()
        return f"❌ Error setting profile: {str(e)}"

def teacher_generate_video(question, script_text, side_img):
    """Generate teaching video using teacher profile"""
    global last_generated_video

    try:
        if not check_access("teacher"):
            return "❌ Access denied. Teacher login required.", None

        print("🎬 Starting video generation...")
        print(f"   Question: {question}")
        print(f"   Script length: {len(script_text) if script_text else 0}")
        print(f"   Side image: {type(side_img)}")

        # Check if profile is set
        face_path = teacher_profile.get("face")
        voice_path = teacher_profile.get("audio")

        if not face_path or not os.path.exists(face_path):
            return "❌ Please set up your teacher profile first (upload face and voice, then click 'Set Profile')", None

        if not question or not question.strip():
            return "❌ Please enter a question for this video", None, None

        script = (script_text or "").strip()
        if not script:
            if voice_path and whisper_model and os.path.exists(voice_path):
                try:
                    print("🎤 Transcribing voice sample...")
                    result = whisper_model.transcribe(voice_path)
                    script = result["text"].strip()
                    print(f"📝 Transcribed: {script[:100]}...")
                except Exception as e:
                    print(f"❌ Transcription error: {str(e)}")
                    return f"❌ Failed to transcribe voice: {str(e)}", None
            else:
                return "❌ Please provide a script or ensure voice sample is uploaded for transcription", None

        # Generate TTS audio
        print("🔊 Generating speech...")
        audio_path = safe_tts_synthesis(script, voice_path)
        if not audio_path:
            return "❌ Failed to generate speech audio", None

        # Generate talking video
        print("🎬 Creating talking video...")
        main_video = generate_talking_video_safe(face_path, audio_path)
        if not main_video:
            return "❌ Failed to generate talking video. Check SadTalker installation.", None, None

        final_video = main_video

        # Add side image if provided
        if side_img is not None:
            print("🖼️ Adding side diagram...")
            side_video = os.path.join(TMP_DIR, f"with_side_{int(time.time())}_{uuid.uuid4().hex[:8]}.mp4")
            fast_side_image_stack(final_video, side_img, side_video)

            if os.path.exists(side_video) and has_video_stream(side_video):
                final_video = side_video
                print("✅ Side diagram added successfully")

        # Add subtitles
        print("📝 Adding subtitles...")
        subtitled_video = os.path.join(TMP_DIR, f"final_{int(time.time())}_{uuid.uuid4().hex[:8]}.mp4")
        add_subtitles(final_video, script, subtitled_video)

        if os.path.exists(subtitled_video) and has_video_stream(subtitled_video):
            final_video = subtitled_video

        # Save to knowledge base
        question_key = question.strip().lower()
        knowledge_base[question_key] = {
            "answer": script,
            "video": final_video,
            "question": question,
            "keywords": question_key.split()
        }
        save_kb()

        # Add to video library
        video_info = {
            "id": len(video_library) + 1,
            "question": question,
            "script": script,
            "timestamp": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
            "concept_diagram": side_img is not None,
            "type": "generated",
            "video_path": final_video
        }
        video_library.append(video_info)

        # Set as last generated
        last_generated_video = final_video

        success_msg = f"✅ Video generated successfully!\n\nQuestion: {question}\n\nScript: {script[:200]}{'...' if len(script) > 200 else ''}\n\nFeatures: {'Side diagram + ' if side_img else ''}Subtitles + Voice synthesis"

        print("🎉 Video generation completed successfully!")
        return success_msg, final_video, gr.update(value=final_video)

    except Exception as e:
        error_msg = f"❌ Video generation error: {str(e)}"
        print(f"❌ Generation error: {str(e)}")
        traceback.print_exc()
        return error_msg, None

# =============================================================================
# LEARNING MATERIAL MANAGEMENT - WITH FILE UPLOAD
# =============================================================================

def extract_text_from_file(file_path):
    """Extract text from uploaded PPT or DOC files"""
    if not file_path:
        return ""

    try:
        if file_path.endswith('.docx') and docx:
            doc = docx.Document(file_path)
            text = '\n'.join([paragraph.text for paragraph in doc.paragraphs])
            return text
        elif file_path.endswith('.pptx') and Presentation:
            prs = Presentation(file_path)
            text = []
            for slide in prs.slides:
                for shape in slide.shapes:
                    if hasattr(shape, "text"):
                        text.append(shape.text)
            return '\n'.join(text)
        else:
            return "Unsupported file format. Please use .docx or .pptx files."
    except Exception as e:
        return f"Error reading file: {str(e)}"

def handle_script_file_upload(file):
    """Handle script file upload for teaching script"""
    if file is None:
        return ""

    extracted_text = extract_text_from_file(file.name)
    return extracted_text if extracted_text else ""

def handle_multiple_notes_files(files):
    """Handle multiple files and combine content"""
    if not files:
        return lesson_content["notes"]

    combined_content = []
    for file_obj in files:
        if file_obj:
            file_path = file_obj.name if hasattr(file_obj, 'name') else str(file_obj)
            text = extract_text_from_file(file_path)
            if text and not text.startswith("Error"):
                filename = file_path.split('/')[-1]
                combined_content.append(f"\n📄 **{filename}**\n{'-'*50}\n{text}")


def create_teacher_file_sharing_tab():
    """Teacher file sharing interface - REPLACES text extraction"""

    with gr.Column():


        with gr.Row():
            file_upload = gr.File(
                label="📁 Select File to Share",
                file_types=[".pdf", ".docx", ".pptx", ".txt", ".png", ".jpg", ".jpeg", ".xlsx"],
                type="filepath"
            )

            category_select = gr.Dropdown(
                choices=["notes", "assignments", "resources"],
                value="notes",
                label="📂 Category"
            )



        share_btn = gr.Button("📤 Share File with Students", variant="primary")
        share_status = gr.Markdown("")

        # Show currently shared files
        refresh_btn = gr.Button("🔄 Refresh Files", variant="secondary")
        gr.Markdown("### 📋 Currently Shared Files")
        shared_files_display = gr.Markdown("")


        def handle_file_share(uploaded_file, category, description):
            if not uploaded_file:
                return "⚠️ Please select a file to share", ""

            result = share_file_with_students(uploaded_file, category, description)
            files_display = get_current_files_display()
            return result, files_display

        def get_current_files_display():
            files = get_shared_files()
            if not files:
                return "📭 No files shared yet"

            display = f"**📊 Total Files Shared: {len(files)}**\n\n"


            return display

        # Event handlers
        share_btn.click(
            fn=handle_file_share,
            inputs=[file_upload, category_select],
            outputs=[share_status, shared_files_display]
        )

        refresh_btn.click(
            fn=get_current_files_display,
            outputs=[shared_files_display]
        )

        # Initialize display
        shared_files_display.value = get_current_files_display()

    return file_upload, category_select, share_btn, share_status

def update_notes(notes_text):
    """Update lesson notes"""
    if not check_access("teacher"):
        return "❌ Access denied. Teacher login required."

    lesson_content["notes"] = notes_text
    save_lesson_content()
    return "✅ Notes updated successfully!"

def add_quiz_question(question, option1, option2, option3, option4, correct_answer):
    if not check_access("teacher"):
        return "❌ Access denied. Teacher login required."

    if not all([question, option1, option2, option3, option4, correct_answer]):
        return "❌ Please fill all fields."

    if correct_answer not in [option1, option2, option3, option4]:
        return "❌ Correct answer must be one of the options."

    new_quiz = {
        "question": question,
        "options": [option1, option2, option3, option4],
        "answer": correct_answer
    }

    lesson_content["quiz"].append(new_quiz)
    save_lesson_content()
    load_lesson_content()
    return f"✅ Quiz question added! Total questions: {len(lesson_content['quiz'])}"
def get_quiz_question_choices():
    """Get list of quiz questions for dropdown"""
    if not lesson_content["quiz"]:
        return ["No questions available"]

    choices = []
    for i, q in enumerate(lesson_content["quiz"]):
        question_text = q["question"]
        # Truncate long questions for dropdown display
        display_text = f"Q{i+1}: {question_text[:50]}..." if len(question_text) > 50 else f"Q{i+1}: {question_text}"
        choices.append(display_text)

    return choices

def delete_quiz_question(selected_question):
    """Delete selected quiz question"""
    if not check_access("teacher"):
        return "❌ Access denied. Teacher login required.", gr.update()

    if not selected_question or selected_question == "No questions available":
        return "❌ Please select a question to delete.", gr.update()

    try:
        # Extract question number from dropdown selection (e.g., "Q1:" -> 0)
        if selected_question.startswith("Q") and ":" in selected_question:
            question_num = int(selected_question.split(":")[0][1:]) - 1  # Q1 -> 0, Q2 -> 1, etc.

            if 0 <= question_num < len(lesson_content["quiz"]):
                deleted_question = lesson_content["quiz"].pop(question_num)
                save_lesson_content()

                # Return success message and updated dropdown choices
                success_msg = f"✅ Question deleted: {deleted_question['question'][:50]}..."
                updated_choices = get_quiz_question_choices()

                return success_msg, gr.update(choices=updated_choices, value=None)
            else:
                return "❌ Invalid question number.", gr.update()
        else:
            return "❌ Invalid selection format.", gr.update()

    except Exception as e:
        return f"❌ Error deleting question: {str(e)}", gr.update()

def delete_quiz_and_refresh_all(selected_question):
    """Delete quiz question and refresh all components"""
    delete_msg, dropdown_update = delete_quiz_question(selected_question)

    # Refresh student components with updated quiz
    _, quiz_upd, _, _, _, _, _ = refresh_student_components()

    # Update quiz preview for teacher
    quiz_preview = f"Total Quiz Questions: {len(lesson_content['quiz'])}\n\n" + \
                  "\n".join([f"Q{i+1}: {q['question']}\nAnswer: {q['answer']}\n"
                            for i, q in enumerate(lesson_content['quiz'])])

    return delete_msg, dropdown_update, gr.update(value=quiz_preview), gr.update(value=quiz_preview)

# =============================================================================
# STUDENT FUNCTIONS
# =============================================================================
def improved_knowledge_search(question):
    """Enhanced knowledge base search with better topic differentiation"""
    if not question or not question.strip():
        return None, None

    query = question.strip().lower()

    # Remove common question words to focus on the topic
    question_words = ["what", "is", "are", "how", "why", "when", "where", "define", "explain"]
    query_words = [word for word in query.split() if word not in question_words and len(word) > 2]

    print(f"🔍 Searching for: '{query}' -> Keywords: {query_words}")

    # 1. Exact match
    if query in knowledge_base:
        entry = knowledge_base[query]
        print(f"✅ Exact match found: {query}")
        return entry.get("video"), entry.get("answer")

    # 2. Key topic match
    for key, entry in knowledge_base.items():
        if key.startswith("library_"):
            continue
        for word in query_words:
            if key == word or key.startswith(word + " ") or key.endswith(" " + word):
                print(f"✅ Topic match found: '{word}' -> '{key}'")
                return entry.get("video"), entry.get("answer")

    # 3. Partial match
    for key, entry in knowledge_base.items():
        if key.startswith("library_"):
            continue
        if query in key or key in query:
            print(f"✅ Partial match found: {key}")
            return entry.get("video"), entry.get("answer")

    # 4. Scored word-based match
    best_match = None
    best_score = 0
    best_key = ""

    for key, entry in knowledge_base.items():
        if key.startswith("library_"):
            continue

        stored_keywords = entry.get("keywords", key.split())
        key_words = key.split()

        score = 0
        matched_words = []

        for word in query_words:
            if word in stored_keywords:
                score += 3
                matched_words.append(word)
            elif word in key_words:
                score += 2
                matched_words.append(word)
            else:
                for stored_word in stored_keywords + key_words:
                    if word in stored_word or stored_word in word:
                        score += 1
                        matched_words.append(f"{word}~{stored_word}")
                        break

        if len(matched_words) > 1:
            score += len(matched_words)

        if score >= 2 and score > best_score:
            best_score = score
            best_match = entry
            best_key = key
            print(f"🎯 Better match: '{key}' (score: {score}, words: {matched_words})")

    if best_match and best_score >= 2:
        print(f"✅ Best match selected: '{best_key}' (final score: {best_score})")
        return best_match.get("video"), best_match.get("answer")

    print(f"❌ No suitable match found for: {query}")
    return None, None

def create_simple_downloads():
    """Simple file downloads that actually work"""

    with gr.Column():
        gr.Markdown("### 📥 Download Course Files")

        # All files in one place
        file_dropdown = gr.Dropdown(
            label="📁 Select File to Download",
            choices=[],
            value=None
        )

        download_btn = gr.Button("⬇️ Download Selected File", variant="primary")
        downloaded_file = gr.File(label="Your Downloaded File")

        refresh_btn = gr.Button("🔄 Refresh File List")
        files_info = gr.Markdown("Click refresh to see available files...")

        def get_all_files_for_download():
            """Get all files for download dropdown"""
            all_files = get_shared_files()  # Get all files

            if not all_files:
                return [], "📭 No files available yet"

            choices = []
            info_text = f"**📊 Total Available Files: {len(all_files)}**\n\n"

            for file_record in all_files:
                file_id = file_record.get("id")
                file_name = file_record.get("original_name")
                category = file_record.get("category", "unknown")
                file_size = file_record.get("file_size")

                # Format: "filename (category) - size"
                display_name = f"{file_name} ({category.title()}) - {file_size}"
                choices.append((display_name, file_id))



            return choices, info_text

        def download_file(file_id):
            """Download the selected file"""
            if not file_id:
                return None

            try:
                all_files = get_shared_files()
                for file_record in all_files:
                    if file_record.get("id") == file_id:
                        file_path = file_record.get("file_path")
                        if file_path and os.path.exists(file_path):
                            print(f"📥 Downloading: {file_record.get('original_name')}")
                            return file_path
                return None
            except Exception as e:
                print(f"❌ Download error: {e}")
                return None

        def refresh_files():
            """Refresh the file list"""
            choices, info = get_all_files_for_download()
            return gr.update(choices=choices), info

        # Event handlers
        refresh_btn.click(
            fn=refresh_files,
            outputs=[file_dropdown, files_info]
        )

        download_btn.click(
            fn=download_file,
            inputs=[file_dropdown],
            outputs=[downloaded_file]
        )

        # Initialize
        choices, info = get_all_files_for_download()
        file_dropdown.choices = choices
        files_info.value = info

    return file_dropdown, downloaded_file, files_info

def handle_file_download(file_id):
    """Handle file download request"""
    try:
        files = get_shared_files()
        for file_record in files:
            if file_record.get("id") == file_id:
                file_path = file_record.get("file_path")
                if file_path and os.path.exists(file_path):
                    print(f"📥 Serving download: {file_record.get('original_name')}")
                    return file_path
        return None
    except Exception as e:
        print(f"❌ Download error: {e}")
        return None





def student_query_flow(question):
    """Handle student questions"""
    if not check_access("student"):
        return None, "❌ Please log in to ask questions."

    try:
        if not question or not question.strip():
            return None, "❌ Please enter a question."
        add_question_to_history(question)
        # Search existing knowledge
        found_video, found_answer = improved_knowledge_search(question)
        if found_video and found_answer:
            if os.path.exists(found_video):
                return found_video, found_answer

        # Generate new response
        face_path = teacher_profile.get("face")
        voice_path = teacher_profile.get("audio")

        if not face_path or not os.path.exists(face_path):
            return None, "❌ Teacher profile not set. Please ask teacher to set up profile first."

        answer_text = safe_llm_answer(question)

        audio_path = safe_tts_synthesis(answer_text, voice_path)
        if not audio_path:
            return None, "❌ Failed to generate response audio."

        video_path = generate_talking_video_safe(face_path, audio_path)
        if not video_path:
            return None, "❌ Failed to generate response video."

        # Save to knowledge base
        question_key = question.strip().lower()
        knowledge_base[question_key] = {
            "answer": answer_text,
            "video": video_path,
            "question": question,
            "keywords": question_key.split()
        }
        save_kb()

        return video_path, answer_text

    except Exception as e:
        print(f"❌ Student query error: {str(e)}")
        return None, f"❌ Error processing question: {str(e)}"

# =============================================================================
# VIDEO LIBRARY FUNCTIONS
# =============================================================================

def get_video_options():
    """Get available videos for dropdown"""
    try:
        options = []
        for key, entry in knowledge_base.items():
            if not key.startswith("library_") and "video" in entry:
                video_path = entry["video"]
                if os.path.exists(video_path):
                    question = entry.get("question", key)
                    display_text = f"Q: {question[:60]}..." if len(question) > 60 else f"Q: {question}"
                    options.append(display_text)

        return options if options else ["No videos available"]
    except Exception as e:
        print(f"Error getting video options: {e}")
        return ["Error loading videos"]

def play_selected_video(selected_option):
    """Play selected video from dropdown"""
    if not selected_option or selected_option in ["No videos available", "Error loading videos"]:
        return None

    try:
        for key, entry in knowledge_base.items():
            if "video" in entry and os.path.exists(entry["video"]):
                question = entry.get("question", entry.get("info", key))
                if question in selected_option:
                    return entry["video"]
        return None
    except Exception as e:
        print(f"Error playing video: {e}")
        return None

def refresh_video_dropdown():
    """Refresh video dropdown"""
    return gr.update(choices=get_video_options(), value=None)

# =============================================================================
# MAIN APPLICATION - FIXED LAYOUT
# =============================================================================

def create_complete_app():
    """Create the complete GURU Virtual Tutor application with FIXED LAYOUT"""

    with gr.Blocks(css=custom_css, fill_height=True, title="🎓 GURU - AI Virtual Tutor") as app:

        # State management
        login_state = gr.State(False)

        # =============================================================================
        # SPLASH SCREEN
        # =============================================================================
        with gr.Group(visible=True) as splash_screen:
            gr.HTML(f"""
            <div class="splash-screen">
                <div class="logo-container">
                    {create_logo()}
                    <div class="spinner"></div>
                    <h2 style="color: #1e40af; margin-top: 25px; font-weight: 700; font-size: 1.6em;">Loading GURU - A Virtual Tutor...</h2>
                    <p style="color: #64748b; margin-top: 10px; font-size: 1.1em;">AI-Powered Learning Platform</p>
                </div>
            </div>
            """)

        # =============================================================================
        # LOGIN SCREEN
        # =============================================================================
        with gr.Group(visible=False) as login_screen:
            gr.HTML("<br><br>")
            with gr.Row():
                with gr.Column(scale=1):
                    pass
                with gr.Column(scale=2):
                    gr.HTML(f"""
                    <div class="login-container">
                        {create_logo()}
                        <h2 style="text-align: center; color: #1e40af; margin-bottom: 35px; font-weight: 700; font-size: 1.6em;">Welcome to GURU - A Virtual Tutor</h2>
                    """)

                    username_input = gr.Textbox(
                        label="👤 Username",
                        placeholder="Enter your username here...",
                        elem_classes=["login-input"]
                    )
                    password_input = gr.Textbox(
                        label="🔒 Password",
                        type="password",
                        placeholder="Enter your password here...",
                        elem_classes=["login-input"]
                    )
                    login_btn = gr.Button(
                        "🚀 Sign In",
                        variant="primary",
                        elem_classes=["primary-button"]
                    )
                    login_message = gr.HTML()

                    gr.HTML("""
                    <div style="background: linear-gradient(135deg, #f0f9ff 0%, #e0f2fe 100%); padding: 25px; border-radius: 15px; margin-top: 25px; border: 2px solid rgba(59, 130, 246, 0.2);">
                        <h4 style="color: #1e40af; margin-bottom: 15px; font-size: 1.3em;">📝 Demo Credentials</h4>
                        <p style="color: #1e293b; margin: 8px 0; font-size: 1.1em;"><strong>👩‍🏫 Teacher:</strong> username: <code style="background: #dbeafe; padding: 2px 6px; border-radius: 4px;">teacher</code>, password: <code style="background: #dbeafe; padding: 2px 6px; border-radius: 4px;">teach123</code></p>
                        <p style="color: #1e293b; margin: 8px 0; font-size: 1.1em;"><strong>🎓 Student:</strong> username: <code style="background: #dbeafe; padding: 2px 6px; border-radius: 4px;">student</code>, password: <code style="background: #dbeafe; padding: 2px 6px; border-radius: 4px;">learn123</code></p>
                        <p style="color: #1e293b; margin: 8px 0; font-size: 1.1em;"><strong>👨‍💼 Admin:</strong> username: <code style="background: #dbeafe; padding: 2px 6px; border-radius: 4px;">admin</code>, password: <code style="background: #dbeafe; padding: 2px 6px; border-radius: 4px;">admin123</code></p>
                    </div>
                    </div>""")
                with gr.Column(scale=1):
                    pass

        # =============================================================================
        # TEACHER DASHBOARD - FIXED LAYOUT AS REQUESTED
        # =============================================================================
        with gr.Group(visible=False) as teacher_dashboard:
            gr.HTML(f"""
            <div class="app-header">
                <h1 class="app-title">🎓 E-Education System</h1>
                <div class="app-subtitle">👩‍🏫 Teacher Dashboard - Create & Manage Content</div>
                <div style="margin-top: 20px;">
                    <span style="background: linear-gradient(135deg, #10b981 0%, #059669 100%); color: white; padding: 8px 16px; border-radius: 20px; font-weight: 600;">
                        Welcome, {get_welcome_name("Teacher")}!
                    </span>
                    <button class="logout-button" onclick="location.reload()" style="margin-left: 15px;">🚪 Logout</button>
                </div>
            </div>
            """)

            with gr.Row(equal_height=True):
                # LEFT PANEL - Avatar Setup & Video Generation (Orange theme)
                with gr.Column(scale=1, elem_classes=["left-panel"]):
                    gr.HTML('<div class="section-header">📝 Avatar & Video Generation</div>')

                    # Avatar Setup
                    gr.HTML('<h3 style="color: #b45309; margin: 20px 0 12px 0; font-weight: 800; font-size: 1.5em;">🔧 Avatar Setup</h3>')

                    avatar_img_upload = gr.Image(
                        label="📸 Upload Avatar Image",
                        type="pil",
                        height=260,
                        interactive=True
                    )
                    avatar_voice_upload = gr.Audio(
                        label="🎤 Upload Avatar Voice",
                        type="filepath",
                        interactive=True
                    )
                    set_profile_btn = gr.Button("💾 Set as Profile", elem_classes=["upload-button"])
                    profile_status = gr.HTML()

                    # Video Generation
                    gr.HTML('<h3 style="color: #b45309; margin: 25px 0 12px 0; font-weight: 800; font-size: 1.5em;">🎬 Video Generation</h3>')

                    question_input = gr.Textbox(
                        label="❓ Name Your Script",
                        lines=3,
                        placeholder="What question will this video answer?"
                    )

                    script_file_upload = gr.File(
                        label="📄 Upload PPT/DOC for Script (Optional)",
                        file_types=[".pptx", ".docx"]
                    )

                    teaching_script = gr.Textbox(
                        label="📝 Teaching Script",
                        lines=6,
                        placeholder="Enter your teaching content here...",
                        interactive=True
                    )

                    concept_diagram = gr.Image(
                        label="📋 Concept Diagram (Optional)",
                        type="pil",
                        height=160
                    )

                    generate_btn = gr.Button("🚀 Generate Video", elem_classes=["primary-button"])
                    generation_status = gr.HTML()

                # CENTER PANEL - Learning Materials at TOP, Video Library below (Green theme)

                with gr.Column(scale=1, elem_classes=["center-panel"]):
                    gr.HTML('<div class="section-header">📚 Learning Materials</div>')

                    # Video Library BELOW Learning Materials
                    gr.HTML('<h3 style="color: #047857; margin: 25px 0 12px 0; font-weight: 800; font-size: 1.5em;">📺 Library</h3>')

                    # Generated Video Display
                    teacher_video_display = gr.Video(label="📹 Generated Video", height=260)
                    gr.HTML("<div style='height: 25px;'></div>")
                    # Video Library Dropdown
                    teacher_video_dropdown = gr.Dropdown(
                        label="🎬 Select Video to Play",
                        choices=get_video_options(),
                        interactive=True
                    )
                    with gr.Column():
                       with gr.Row(equal_height=True):
                          play_teacher_video_btn = gr.Button("▶️ Play", elem_classes=["upload-button"], scale=1)

                       with gr.Row(equal_height=True):
                          refresh_teacher_dropdown_btn = gr.Button("🔄 Refresh", elem_classes=["upload-button"], scale=1)
                          download_teacher_video_btn = gr.DownloadButton("⬇️ Download", label="Download", elem_classes=["upload-button"], scale=1)


                    gr.HTML('<h3 style="color: #047857; margin: 25px 0 12px 0; font-weight: 800; font-size: 1.3em;">📘 Learning Material Management</h3>')

                    # Learning Materials at TOP
                    with gr.Tabs():
                        with gr.Tab("📤 Share Files"):
                           gr.Markdown("📂 Upload and share files with students.")
                           file_upload, category_select, share_btn, share_status = create_teacher_file_sharing_tab()
                           load_files_btn = gr.Button("", elem_classes=["upload-button"])
                        with gr.Tab("📝 Notes"):
                            gr.Markdown("✏️ Here you can create, edit, and update lesson notes. ")
                            notes_editor = gr.Textbox(
                               label="Edit Notes (Visible in Notes tab)",
                               value=lesson_content["notes"],
                               lines=6,
                               interactive=True
                            )
                            update_notes_btn = gr.Button("💾 Update Notes", elem_classes=["upload-button"])
                            notes_status = gr.HTML()
                        with gr.Tab("❓ Quiz"):
                            quiz_question = gr.Textbox(label="Question", interactive=True)
                            with gr.Row():
                                quiz_option1 = gr.Textbox(label="Option 1", interactive=True)
                                quiz_option2 = gr.Textbox(label="Option 2", interactive=True)
                            with gr.Row():
                                quiz_option3 = gr.Textbox(label="Option 3", interactive=True)
                                quiz_option4 = gr.Textbox(label="Option 4", interactive=True)
                            quiz_answer = gr.Dropdown(
                                label="Correct Answer",
                                choices=[],
                                interactive=True
                             )
                            add_quiz_btn = gr.Button("➕ Add Question", elem_classes=["upload-button"])
                            quiz_status = gr.HTML()

                            # DELETE SECTION - ADD THIS
                            gr.HTML('<hr style="border: 1px solid rgba(0,0,0,0.1); margin: 20px 0;">')
                            gr.HTML('<h4 style="color: #dc2626; margin: 15px 0 10px 0; font-weight: 800;">🗑️ Delete Question</h4>')

                            quiz_delete_dropdown = gr.Dropdown(
                                label="Select Question to Delete",
                                choices=get_quiz_question_choices(),
                                interactive=True,
                                value=None
                              )

                            delete_quiz_btn = gr.Button("🗑️ Delete Selected Question",
                                elem_classes=["logout-button"],  # Red button style
                                variant="secondary")

                            delete_quiz_status = gr.HTML()


                # RIGHT PANEL - Student Corner (Blue theme) - FIXED LAYOUT AS REQUESTED
                with gr.Column(scale=1, elem_classes=["right-panel"], visible=False):
                    gr.HTML('<div class="section-header">🎓 Student Corner</div>')

                    # Student Corner at TOP - Display Updated Learning Materials
                    gr.HTML('<h3 style="color: #1d4ed8; margin: 20px 0 12px 0; font-weight: 800; font-size: 1.2em;">📖 Learning Materials</h3>')

                    with gr.Tabs():
                        with gr.Tab("📝"):
                            student_notes_preview = gr.Markdown(lesson_content["notes"])
                        with gr.Tab("❓"):
                            student_quiz_preview = gr.Markdown(f"Total Quiz Questions: {len(lesson_content['quiz'])}\n\n" +
                                                             "\n".join([f"Q{i+1}: {q['question']}\nAnswer: {q['answer']}\n"
                                                                       for i, q in enumerate(lesson_content['quiz'])]))
                        with gr.Tab("📋"):
                            student_assignment_preview = gr.Markdown(lesson_content["assignment"])

                    # AI Video Player BELOW Student Corner
                    gr.HTML('<h3 style="color: #1d4ed8; margin: 20px 0 12px 0; font-weight: 800; font-size: 1.2em;">🤖 Video Player</h3>')
                    teacher_ai_video = gr.Video(label="🤖 Response Video", height=160)

                    # AI Response BELOW AI Video Player
                    gr.HTML('<h3 style="color: #1d4ed8; margin: 20px 0 12px 0; font-weight: 800; font-size: 1.2em;">📝 Teacher Response</h3>')
                    teacher_ai_response = gr.Textbox(
                        label="🤖 Teacher Answer",
                        lines=7,
                        interactive=False,
                        placeholder="Teacher's response will appear here..."
                    )
                    gr.HTML('<h3 style="color: #1d4ed8; margin: 20px 0 12px 0;">📋 Student Questions</h3>')
                    teacher_history_dropdown = gr.Dropdown(
                        label="🔍 Previous Student Questions",
                        choices=get_question_choices(),
                        interactive=True
                    )
                    teacher_use_previous_btn = gr.Button("📋 Use Selected", elem_classes=["upload-button"])
                    # Ask Question BELOW AI Response
                    gr.HTML('<h3 style="color: #1d4ed8; margin: 20px 0 12px 0; font-weight: 800; font-size: 1.2em;">❓ Ask Question</h3>')
                    teacher_question_input = gr.Textbox(
                        label="🤔 Ask Question",
                        lines=8,
                        placeholder="Ask the question ..."
                    )
                    teacher_ask_btn = gr.Button("🚀 Ask Teacher", elem_classes=["primary-button"])


                    with gr.Group():
                     gr.HTML(f"""
                        <div style="
                           background: #e0f2fe;
                           border: 2px solid #90cdf4;
                           border-radius: 12px;
                           padding: 20px;
                           margin-top: 15px;
                           font-size: 1em;
                           color: #1f2937;
                           line-height: 1.6;
                        ">
                          <h3 style="margin-top: 0; color: #2563eb;">📊 Learning Stats</h3>
                          <ul style="padding-left: 20px; margin-top: 10px;">
                             <li><strong>📹 Total Videos:</strong> {len(video_library)}</li>
                             <li><strong>❓ Total Quizzes:</strong> {len(lesson_content["quiz"])}</li>
                             <li><strong>📝 Notes Length:</strong> {len(lesson_content["notes"].split())} words</li>
                          </ul>
                        </div>
                        """)


        # =============================================================================
        # STUDENT DASHBOARD - SAME LAYOUT STRUCTURE
        # =============================================================================
        with gr.Group(visible=False) as student_dashboard:
            gr.HTML(f"""
            <div class="app-header">
                <h1 class="app-title">🎓 E-Education System</h1>
                <div class="app-subtitle">🎓 Student Dashboard - Learn & Explore</div>
                <div style="margin-top: 20px;">
                    <span style="background: linear-gradient(135deg, #3b82f6 0%, #2563eb 100%); color: white; padding: 8px 16px; border-radius: 20px; font-weight: 600;">
                        Welcome, {get_welcome_name("Student")}!
                    </span>
                    <button class="logout-button" onclick="location.reload()" style="margin-left: 15px;">🚪 Logout</button>
                </div>
            </div>
            """)

            with gr.Row(equal_height=True):
                # Left Panel - Learning Materials
                with gr.Column(scale=1, elem_classes=["left-panel"], elem_id="left-panel-container"):
                    gr.HTML('<div class="section-header">📚 Learning Materials</div>')

                    with gr.Tabs():
                        with gr.TabItem("📝 Notes"):
                            student_notes = gr.Markdown(lesson_content["notes"])
                        with gr.TabItem("📥 Downloads"):
                            file_dropdown, downloaded_file, files_info = create_simple_downloads()

                        with gr.TabItem("❓ Quiz"):
                            gr.HTML('<h4 style="color: #b45309; margin-bottom: 15px;">🎯 Interactive Quiz</h4>')
                            quiz_result = gr.HTML()

                            # Generate quiz questions dynamically
                            for i, q in enumerate(lesson_content["quiz"]):
                                with gr.Group():
                                    gr.HTML(f'<p style="font-weight: 800; color: #1e293b; font-size: 1.1em; margin: 15px 0 10px 0;"><strong>Q{i+1}:</strong> {q["question"]}</p>')
                                    quiz_radio = gr.Radio(
                                        choices=q["options"],
                                        label="",
                                        value=None
                                    )

                                    # Create answer checker
                                    def make_check_answer(correct_answer):
                                        def check_answer(selected):
                                            if selected == correct_answer:
                                                return "<div style='color: #16a34a; font-weight: 700; padding: 10px; background: #dcfce7; border-radius: 8px; margin: 10px 0;'>✅ Correct! Well done!</div>"
                                            else:
                                                return f"<div style='color: #dc2626; font-weight: 700; padding: 10px; background: #fecaca; border-radius: 8px; margin: 10px 0;'>❌ Incorrect. The correct answer is: {correct_answer}</div>"
                                        return check_answer

                                    quiz_radio.change(
                                        make_check_answer(q["answer"]),
                                        inputs=[quiz_radio],
                                        outputs=[quiz_result]
                                    )

                        with gr.TabItem("📋 Assignment"):
                            student_assignment = gr.Markdown(lesson_content["assignment"])

                        with gr.TabItem("🔗 Resources"):
                            student_resources = gr.Markdown(lesson_content["resources"])

                        with gr.TabItem("✨ Extras"):
                            student_extras = gr.Markdown(lesson_content["extras"])

                        with gr.TabItem("🃏 Flashcards"):
                            student_flashcards = gr.Markdown(lesson_content["flashcards"])

                        with gr.TabItem("💡 Tips"):
                            student_tips = gr.Markdown(lesson_content["tips"])
                             # Learning Stats

                    gr.HTML('<h3 style="color: #0ea5e9 margin: 25px 0 12px 0; font-weight: 800; font-size: 1.2em;">📊 Learning Stats</h3>')
                    student_stats = gr.HTML(f"""
                    <div style="background: linear-gradient(135deg, #e0f7ff 0%, #b6eaff 100%); padding: 20px; border-radius: 12px; border: 2px solid rgba(14, 165, 233, 0.3);">
                        <h4 style="color:  #0ea5e9; margin-bottom: 15px; font-weight: 800;">📈 Your Progress</h4>
                        <p style="margin: 8px 0;"><strong>Available Videos:</strong> {len([k for k in knowledge_base.keys() if not k.startswith('library_')])}</p>
                        <p style="margin: 8px 0;"><strong>Quiz Questions:</strong> {len(lesson_content.get('quiz', []))}</p>
                        <p style="margin: 8px 0;"><strong>Study Materials:</strong> 6 sections</p>
                        <p style="margin: 8px 0;"><strong>Last Updated:</strong> {datetime.now().strftime('%H:%M:%S')}</p>
                    </div>
                    """)
                    gr.HTML("""
                    <div style="margin-top: 20px; background: #f0f9ff; padding: 18px; border-radius: 12px; border: 2px solid #bae6fd;">
                       <h4 style="color: #0ea5e9; margin-bottom: 12px; font-weight: 800;">📌 Quick Actions</h4>
                       <ul style="padding-left: 20px; margin: 0;">
                         <li>🧠 Review recent notes</li>
                         <li>🎥 Rewatch AI-generated videos</li>
                         <li>📝 Attempt quizzes</li>
                         <li>💬 Ask a follow-up question in doubt section</li>
                         <li>📈 Check your learning stats</li>



                      </ul>
                    </div>
                    """)


                # Center Panel - Video Library
                with gr.Column(scale=1, elem_classes=["center-panel"]):
                    gr.HTML('<div class="section-header">📺 Video Library</div>')

                    gr.HTML('<h3 style="color: #047857; margin: 20px 0 12px 0; font-weight: 800; font-size: 1.2em;">🎬 Video Player</h3>')
                    video_display = gr.Video(label="📹 Lesson Video", height=300,interactive=False,sources=None )
                    video_state = gr.State()  # Store current video












                    gr.HTML('<h3 style="color: #047857; margin: 20px 0 12px 0; font-weight: 800; font-size: 1.2em;">📋 Available Videos</h3>')
                    student_video_dropdown = gr.Dropdown(
                        label="🎬 Browse Videos",
                        choices=get_video_options(),
                        interactive=True
                    )


                    with gr.Column():
                        play_video_btn = gr.Button("▶️ Play Video", elem_classes=["upload-button"], scale=1)
                        refresh_student_dropdown_btn = gr.Button("🔄 Refresh", elem_classes=["upload-button"],scale=1)



                # Right Panel - AI Teacher
                with gr.Column(scale=1, elem_classes=["right-panel"]):
                    gr.HTML('<div class="section-header">🤖 Ask doubt</div>')

                    gr.HTML('<h3 style="color: #1d4ed8; margin: 20px 0 12px 0; font-weight: 800; font-size: 1.2em;">🎬 Video Response</h3>')
                    ai_video_response = gr.Video(label="🤖 Teacher Video", height=220)

                    gr.HTML('<h3 style="color: #1d4ed8; margin: 20px 0 12px 0; font-weight: 800; font-size: 1.2em;">📝 Explanation</h3>')
                    ai_text_response = gr.Textbox(
                        label="🤖 Detailed Answer",
                        interactive=False,
                        lines=6,
                        placeholder="Teacher's explanation will appear here..."
                    )
                    gr.HTML('<h3 style="color: #1d4ed8; margin: 20px 0 12px 0;">📋 Previous Questions</h3>')
                    student_history_dropdown = gr.Dropdown(
                         label="🔍 Recently Asked Questions",
                         choices=get_question_choices(),
                         interactive=True
                    )
                    use_previous_btn = gr.Button("📋 Use Selected", elem_classes=["upload-button"])
                    gr.HTML('<h3 style="color: #1d4ed8; margin: 20px 0 12px 0; font-weight: 800; font-size: 1.2em;">❓ Ask Your Question</h3>')
                    doubt_input = gr.Textbox(
                        label="🤔 Your Question",
                        lines=4,
                        placeholder="Type your question here... Be specific for better answers!"
                    )
                    submit_doubt_btn = gr.Button("🚀 Ask Teacher", elem_classes=["primary-button"])

        # =============================================================================
        # EVENT HANDLERS
        # =============================================================================

        def show_login_after_splash():
            """Show login screen after splash delay"""
            time.sleep(2.5)
            return {
                splash_screen: gr.update(visible=False),
                login_screen: gr.update(visible=True)
            }

        def handle_login(username, password):
            """Handle user login"""
            success, message = authenticate_user(username, password)
            if success:
                is_teacher = current_user["role"] == "teacher"
                is_student = current_user["role"] == "student"
                load_questions_history()
                question_choices = get_question_choices()
                print(f"🔄 Login refresh: {len(question_choices)} questions loaded")
                base_updates = {
                  login_screen: gr.update(visible=False),
                  teacher_dashboard: gr.update(visible=is_teacher),
                  student_dashboard: gr.update(visible=is_student),
                  login_message: gr.update(value=""),
                  student_history_dropdown: gr.update(choices=question_choices),
                  teacher_history_dropdown: gr.update(choices=question_choices)
             }

            # If student login, refresh their components with latest data
                if is_student:
                    notes_upd, quiz_upd, assign_upd, res_upd, extra_upd, flash_upd, tips_upd = refresh_student_components()
                    base_updates.update({
                       student_notes: notes_upd,
                       student_assignment: assign_upd,
                       student_resources: res_upd,
                       student_extras: extra_upd,
                       student_flashcards: flash_upd,
                       student_tips: tips_upd
                 })

                return base_updates
            else:
                return {
                  login_message: gr.update(value=f'<div style="color: #dc2626; font-weight: 600; font-size: 1.1em; padding: 15px; background: #fecaca; border-radius: 8px; margin: 10px 0;">{message}</div>')
              }



        def update_quiz_dropdown_choices(q1, q2, q3, q4):
            """Update quiz dropdown choices"""
            choices = [opt for opt in [q1, q2, q3, q4] if opt and opt.strip()]
            return gr.update(choices=choices)

        def update_learning_materials_preview(notes_text):
            """Update learning materials preview in real-time"""
            return gr.update(value=notes_text)

        # Auto-show login after splash
        app.load(show_login_after_splash, outputs=[splash_screen, login_screen])
        def pause_video(current_video):
            """Pause video by clearing it and showing pause message"""
            pause_msg = """
                 🙋‍♀️ Student raised hand!

            Please ask your question in Ask Doubt , then click 'Resume Video' to continue.
            """

            return (
                None,  # Clear video = "pause"
                current_video,  # Store in state
                gr.update(value=pause_msg, visible=True),  # Show pause message
                gr.update(visible=False),  # Hide raise hand
                gr.update(visible=True),   # Show resume
                "⏸️ Video paused"
            )

        def resume_video(stored_video):
            """Resume video by restoring it"""
            return (
                stored_video,  # Restore video
                stored_video,  # Keep in state
                gr.update(visible=False),  # Hide pause message
                gr.update(visible=True),   # Show raise hand
                gr.update(visible=False),  # Hide resume
                "▶️ Video playing"
            )


        # Event handlers


        # Login events
        login_btn.click(
            handle_login,
            inputs=[username_input, password_input],
            outputs=[login_screen, teacher_dashboard, student_dashboard, login_message,
                 student_notes, student_assignment, student_resources,
                 student_extras, student_flashcards, student_tips, student_history_dropdown, teacher_history_dropdown]
        )


        password_input.submit(
            handle_login,
            inputs=[username_input, password_input],
            outputs=[login_screen, teacher_dashboard, student_dashboard, login_message, student_notes, student_assignment, student_resources, student_extras, student_flashcards, student_tips, student_history_dropdown, teacher_history_dropdown]
        )



        # Question selection (Student)
        use_previous_btn.click(
            select_from_dropdown,
            inputs=[student_history_dropdown],
            outputs=[doubt_input]
        )

        # Question selection (Teacher)
        teacher_use_previous_btn.click(
            select_from_dropdown,
            inputs=[teacher_history_dropdown],
            outputs=[teacher_question_input]
        )
        # Teacher events
        avatar_img_upload.change(upload_avatar_image, inputs=[avatar_img_upload], outputs=[profile_status])
        avatar_voice_upload.change(upload_avatar_voice, inputs=[avatar_voice_upload], outputs=[profile_status])
        set_profile_btn.click(set_profile, outputs=[profile_status])

        script_file_upload.change(handle_script_file_upload, inputs=[script_file_upload], outputs=[teaching_script])

        # ADDED: Notes file upload functionality
       # notes_files_upload.change(handle_multiple_notes_files, inputs=[notes_files_upload], outputs=[notes_editor])
        generate_btn.click(
           teacher_generate_video,
            inputs=[question_input, teaching_script, concept_diagram],
            outputs=[generation_status, teacher_video_display , download_teacher_video_btn]
        )

        # Learning materials with REAL-TIME UPDATES
        def update_notes_and_refresh(notes_text):
          result = update_notes(notes_text)
          notes_upd, _, _, _, _, _, _ = refresh_student_components()
          return result, notes_upd, notes_upd

        update_notes_btn.click(
          update_notes_and_refresh,
          inputs=[notes_editor],
          outputs=[notes_status, student_notes_preview, student_notes]
         )


        for option_input in [quiz_option1, quiz_option2, quiz_option3, quiz_option4]:
            option_input.change(
                update_quiz_dropdown_choices,
                inputs=[quiz_option1, quiz_option2, quiz_option3, quiz_option4],
                outputs=[quiz_answer]
            )

        # Updated add quiz function that also refreshes delete dropdown
        def add_quiz_and_refresh_dropdown(question, option1, option2, option3, option4, correct_answer):
           """Add quiz question and refresh both preview and delete dropdown"""
           result = add_quiz_question(question, option1, option2, option3, option4, correct_answer)

           # Update quiz preview
           quiz_preview = f"Total Quiz Questions: {len(lesson_content['quiz'])}\n\n" + \
                  "\n".join([f"Q{i+1}: {q['question']}\nAnswer: {q['answer']}\n"
                            for i, q in enumerate(lesson_content['quiz'])])

           # Update delete dropdown choices
           updated_choices = get_quiz_question_choices()

           # Refresh student components
           _, quiz_upd, _, _, _, _, _ = refresh_student_components()

           return (result, gr.update(value=quiz_preview), gr.update(value=quiz_preview),
               gr.update(choices=updated_choices, value=None))

        # REPLACE your existing add_quiz_btn.click with this:
        add_quiz_btn.click(
           add_quiz_and_refresh_dropdown,
           inputs=[quiz_question, quiz_option1, quiz_option2, quiz_option3, quiz_option4, quiz_answer],
           outputs=[quiz_status, student_quiz_preview, student_quiz_preview, quiz_delete_dropdown]
         )

        # ADD this new delete event handler:
        delete_quiz_btn.click(
           delete_quiz_and_refresh_all,
           inputs=[quiz_delete_dropdown],
           outputs=[delete_quiz_status, quiz_delete_dropdown, student_quiz_preview, student_quiz_preview]
        )




        # Video library
        play_teacher_video_btn.click(play_selected_video, inputs=[teacher_video_dropdown], outputs=[teacher_video_display])
        refresh_teacher_dropdown_btn.click(refresh_video_dropdown, outputs=[teacher_video_dropdown])

        # Student events
        def submit_with_update(question):
            video, text = student_query_flow(question)
            choices = get_question_choices()
            return video, text, gr.update(choices=choices), gr.update(choices=choices)

        # Teacher AI test
        teacher_ask_btn.click(
            submit_with_update,
            inputs=[teacher_question_input],
            outputs=[teacher_ai_video, teacher_ai_response, student_history_dropdown, teacher_history_dropdown]
        )


        submit_doubt_btn.click(
            submit_with_update,
            inputs=[doubt_input],
            outputs=[ai_video_response, ai_text_response, student_history_dropdown, teacher_history_dropdown]
        )
        play_video_btn.click(play_selected_video, inputs=[student_video_dropdown], outputs=[video_display])
        refresh_student_dropdown_btn.click(refresh_video_dropdown, outputs=[student_video_dropdown])

        # Auto-refresh after questions
        submit_doubt_btn.click(refresh_video_dropdown, outputs=[student_video_dropdown])
        teacher_ask_btn.click(refresh_video_dropdown, outputs=[teacher_video_dropdown])
        # Auto-refresh student components when app loads
        def app_startup():
             load_questions_history()
             return (*refresh_student_components(),
                  gr.update(choices=get_quiz_question_choices()),
                  gr.update(choices=get_question_choices()),
                  gr.update(choices=get_question_choices()))

        app.load(
          app_startup,
           outputs=[student_notes, student_quiz_preview, student_assignment,
            student_resources, student_extras, student_flashcards, student_tips,
               quiz_delete_dropdown, student_history_dropdown, teacher_history_dropdown]
          )

    return app

# =============================================================================
# APPLICATION LAUNCH
# =============================================================================
if __name__ == "__main__":
    print("🚀 Starting GURU Virtual Tutor Platform...")

    # Ensure all directories exist
    os.makedirs(TMP_DIR, exist_ok=True)
    os.makedirs(STATIC_DIR, exist_ok=True)

    app = create_complete_app()

    print("\n🎓 GURU Virtual Tutor is Ready!")


    # Launch application
    app.launch(debug=True, share=True, allowed_paths=["/content/tmp", "/content", "/tmp", "/content/shared_files/notes","/content/shared_files/assignments","/content/shared_files/resources"], server_name="0.0.0.0", server_port=7860)



```
